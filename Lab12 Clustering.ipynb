# Step 1: Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs, make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
```

# Step 2: Generate dummy data using make_blobs

```python
x, y = make_blobs(n_samples=15,centers=3, random_state = 41)
```

# Step 3: Plot Generated data 

```python
plt.scatter(x[::,0],x[::,1])
```

# Step 4: Apply Hierarchical Clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
model = AgglomerativeClustering(n_clusters = 3)
```

```python
model.fit(x)
```

```python
model.labels_
```

# Step 5: Display Dendrogram

```python
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
linkage_data = linkage(x, method = 'ward')
```

```python
dendrogram(linkage_data)
plt.show()
```

# Step 6: Generate dummy data using make_moons

```python
x, y = make_moons(n_samples = 300, noise = 0.1, random_state = 42)
```

```python
x = StandardScaler().fit_transform(X=x)
```

# Step 7: Plot Generated data

```python
plt.scatter(x[:, 0], x[:, 1])
```

# Step 8: apply DBSCAN Clustering 

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan = DBSCAN(eps = 0.2, min_samples = 5)
```

```python
dbscan.fit(x)
```

```python
dbscan.labels_
```

```python
plt.scatter(x[:, 0], x[:, 1], c=dbscan.labels_)
plt.show()
```

# Step 9: apply K Mean Clustering on same data

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters = 3, random_state = 42)
kmeans_labels = kmeans.fit_predict(x)

centerdata = kmeans.cluster_centers_
```

```python
plt.scatter(x[:, 0], x[:, 1])
plt.scatter(centerdata[:, 0], centerdata[:, 1])
plt.show()
```
